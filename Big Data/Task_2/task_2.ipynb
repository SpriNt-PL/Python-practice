```python
raw_data = [
    {"dt": "2026-01-17 10:00:01", "lvl": "INFO", "msg": "Login", "user": 101},
    {"dt": "2026-01-17 10:05:22", "lvl": "ERROR", "msg": "DB Error", "user": None}, # Brak usera!
    {"dt": "2026-01-17 10:07:01", "lvl": "INFO", "msg": "Upload", "user": 102},
    {"dt": "2026-01-18 08:00:00", "lvl": "ERROR", "msg": "Timeout", "user": 101},
    {"dt": "2026-01-18 09:15:00", "lvl": "INFO", "msg": "Login", "user": 105}
]
```

```python
import pandas as pd

df = pd.DataFrame(raw_data)

df['dt'] = pd.to_datetime(df['dt'])

df['lvl'] = df['lvl'].astype("category")

display(df.head())
display(df.info())

number_of_na = sum(df['user'].isna())
number_of_na

df['user'].fillna(0, inplace=True)

display(df.head())

date_occurances = df['dt'].dt.date.value_counts()
print(date_occurances)

df_errors = df['lvl']
df_errors.to_csv("error.csv", index=False)

```
