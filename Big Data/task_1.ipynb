Zadanie: "Mini-Analizator Logów Serwera"
Wyobraź sobie, że serwer generuje tysiące linii logów. Twoim zadaniem jest przetworzenie tych surowych danych, aby dowiedzieć się, co dzieje się z systemem.

Twoje zadania:
ETAP 1: Parsowanie (Transform) Przetwórz tę listę stringów na listę słowników (structured data). Każdy słownik powinien mieć klucze: timestamp, level, message, details. Podpowiedź: Użyj metody .split(" | ").

ETAP 2: Agregacja (Analysis) Stwórz słownik statystyk, który policzy, ile razy wystąpił dany poziom logów (INFO, ERROR, WARNING). Oczekiwany wynik: {'INFO': 3, 'ERROR': 2, 'WARNING': 1}.

ETAP 3: Filtrowanie Wyodrębnij tylko logi o poziomie ERROR do osobnej listy i wyświetl same ich wiadomości oraz szczegóły.

ETAP 4: Unikalność Wypisz listę unikalnych wiadomości (messages), które pojawiły się w logach. Podpowiedź: Użyj typu danych set (zbiór).

```python
raw_logs = [
    "2026-01-17 10:00:01 | INFO | User logged in | user_id: 101",
    "2026-01-17 10:05:22 | ERROR | Connection lost | db_host: 10.0.0.5",
    "2026-01-17 10:07:01 | INFO | File uploaded | file_name: report.pdf",
    "2026-01-17 10:10:00 | WARNING | High CPU usage | load: 85%",
    "2026-01-17 10:15:00 | ERROR | Connection lost | db_host: 10.0.0.5",
    "2026-01-17 10:20:05 | INFO | User logged out | user_id: 101"
]
```

```python
import pandas as pd


timestamp = []
level = []
message = []
details = []

for row in raw_logs:
    log = row.split(' | ')
    timestamp.append(log[0])
    level.append(log[1])
    message.append(log[2])
    details.append(log[3])


logs_df = pd.DataFrame({'timestamp': timestamp, 'level': level, 'message': message, 'details' :details})
display(logs_df)    
print(logs_df.info())

# Printing number of appearence of particular level
count_info = logs_df['level'].value_counts()
display(count_info)

error_logs = logs_df[logs_df['level'] == 'ERROR']
error_logs = error_logs[['message', 'details']]
display(error_logs)


unique_messages = logs_df['message'].unique()
print(unique_messages)

    
```
